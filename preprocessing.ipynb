## Preprocessing:

```python
import os

def process_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()

    processed_lines = []
    for line in lines:
        values = line.strip().split()
        if len(values) > 1:  # Check that the line containe more than one value
            # Iterate from the end of the line
            for i in range(len(values) - 1, 0, -1):
                if values[i] == '0':
                    # Set -1 and -2 values to zero
                    if i - 1 > 0:
                        values[i - 1] = '0.000'
                    if i - 2 > 0:
                        values[i - 2] = '0.000'
            processed_lines.append(' '.join(values) + '\n')

    # Write the changes back to the file
    with open(file_path, 'w') as file:
        file.writelines(processed_lines)

def process_folder(folder_path):
    for root, _, files in os.walk(folder_path):
        for file in files:
            if file.endswith('.txt'):
                file_path = os.path.join(root, file)
                process_file(file_path)

# Specify the path to the folder with files
# folder_path = 'dataset/labels/val'
folder_path = 'dataset/labels/train'
process_folder(folder_path)
```
