## Model training:

```python
import os
import mlflow
from ultralytics import YOLO

mlflow_dir = r"C:\Users\andre\Documents\TSI\Degree\PoseSkateTrain\runs\mlflow" # Should end with "\runs\mlflow"
mlflow_uri = "file:///" + mlflow_dir.replace("\\", "/")

os.environ["MLFLOW_TRACKING_URI"] = mlflow_uri
os.environ["MLFLOW_EXPERIMENT_NAME"] = "keypoints-detector"

mlflow.set_tracking_uri(mlflow_uri)

model = YOLO("yolo11s-pose.pt")

results = model.train(
    data="./dataset/data.yaml",
    epochs=200,
    imgsz=640,
    batch=16,
    workers=0,
    project=r"C:\Users\andre\Documents\TSI\Degree\PoseSkateTrain\runs\pose", # Should end with "\runs\pose"
)
```
